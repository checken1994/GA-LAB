# Continual Learning Pipeline (RL-HF)

This notebook demonstrates how to load the `learning_buffer` and prepare it for fine-tuning.

```python
import json
from pathlib import Path

buffer_dir = Path('../learning_buffer')
records = []
if buffer_dir.exists():
    for f in buffer_dir.glob('*.json'):
        with open(f) as file:
            records.append(json.load(file))
print(f"Loaded {len(records)} records for fine-tuning.")
```
